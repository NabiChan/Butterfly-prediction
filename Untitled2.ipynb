```python
device=torch.device("cuda" if torch.cuda.is_available() else 'cpu')
device
```

```python
train_transform =T.Compose([T.Resize(512),
                      T.ToTensor(),
                      T.RandomHorizontalFlip(p=0.5),
                      T.Normalize((0.485, 0.456, 0.406),(0.229, 0.224, 0.225))])
```

```python
train_path= ImageFolder(root=train_img,transform=train_transform)
train_path[0][0].shape
```

```python
train_loader = dataloader.DataLoader(train_path, shuffle=True, batch_size=32)
test_loader = dataloader.DataLoader(test_img, batch_size=32)
```

```python
class BruhNN(nn.Module):
    def __init__(self,color_channel,classes):
        super(BruhNN, self).__init__()
        self.Convlayer= nn.Sequential(
            nn.Conv2d(color_channel,64,kernel_size=3,padding =1),
            nn.BatchNorm2d(64),
            nn.LeakyReLU(),
            nn.MaxPool2d(kernel_size=3,stride=2),
            
            nn.Conv2d(64,128,kernel_size=3,padding =1),
            nn.BatchNorm2d(128),
            nn.LeakyReLU(),
            nn.MaxPool2d(kernel_size=3,stride=2),
            nn.AdaptiveAvgPool2d((4, 4))
        )
        self.flattenlayer= nn.Sequential(
            nn.Flatten(),
            nn.Linear(128*4*4,512),
            nn.LeakyReLU(),
            nn.Dropout(0.5),
            nn.Linear(512,classes)
        )
    def forward(self,x):
        return self.flattenlayer(self.Convlayer(x))
```

```python
model= BruhNN(3,75)
```

```python
optimizer=torch.optim.AdamW(model.parameters(),lr=0.01)
loss=nn.CrossEntropyLoss()
wrong_stored=[]
model.to(device)
for epoch in range(10):
    model.train()
    train_loss=[]
    for i,(inputs,labels) in enumerate(train_loader):
        optimizer.zero_grad()
        predicted=model(inputs.to(device))
        wrong=loss(predicted,labels.to(device))
        wrong.backward()
        optimizer.step()
        train_loss.append(wrong)
        
    with torch.no_grad():
        model.eval()
        for inputs,labels in test_loader:
            y_pred = model(inputs.to(device))
            wrong_test=loss(y_pred,labels.to(device))
        
            avg_test_loss = sum(wrong_test) / len(wrong_test)
            wrong_stored.append(avg_train_loss)
            print('epoch {}, test loss {}'.format(epoch, avg_test_loss)) 


wrong_stored
```
